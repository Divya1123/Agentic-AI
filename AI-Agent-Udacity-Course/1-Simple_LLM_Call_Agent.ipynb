```python
from openai import OpenAI
from dotenv import load_dotenv
import os
```

```python
load_dotenv('config.env')
api_key = os.getenv('OPENAI_API_KEY')
```

```python
client = OpenAI(api_key = api_key)
```

```python
system_prompt = "You are a senior Python Developer. You don't have knowledge about any other language, so don't provide answer for any other language."
user_query = "What is Java Virtual Machine?"
```

```python
messages = [
    {
        "role": "system",
        "content": system_prompt
    },
    {
        "role": "user",
        "content": user_query
    }
]

response = client.chat.completions.create(
    model = 'gpt-4o-mini',
    messages = messages,
    temperature = 0.3
)

response = response.choices[0].message.content

print(response)
```

## 3. creating Memory Layer
- To make it more efficient, we will abstract it into memory class

```python
class Memory:
    def __init__(self):
        self.messages = []

    def add_message(self, role, content):
        self.messages.append(
            {
                "role": role,
                "content": content
            }
        )
    
    def get_messages(self):
        return self.messages
```

```python
def chat(user_query=None, memory=None):
    messages = [{"role": "user", "content": user_query}]

    if memory:
        if user_query:
            memory.add_message(role="user", content = user_query)
        messages = memory.get_messages()

    response = client.chat.completions.create(
        model = 'gpt-4o-mini',
        messages = messages,
        temperature = 0.3
    )

    ai_message = response.choices[0].message.content
    if memory:
        memory.add_message(role="assistant", content = ai_message)
    
    return ai_message
```

```python
memory = Memory()

memory.add_message(role='system', content='You are helpful assistant')

memory.get_messages()
```

```python
result = chat(user_query = "What's capital of France?")
print(result)
```

```python
memory.get_messages()
```

```python
response = chat(user_query="What is capital of France", memory= memory)

print(response)
print('\n\n', memory.get_messages())
```
